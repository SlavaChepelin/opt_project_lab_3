## Выводы

1. Плато
Для обоих датасетов плато наступает при $l \approx 50$: дальнейшее увеличение $l$ не сокращает число итераций.
2. Золотая середина
Оптимальное $l$ по времени работы – **10–20**. При $l>50$ время растёт из-за усложнения итерации $O(l\cdot n)$, хотя итераций становится не меньше. Для `a9a` минимальное время достигнуто при $l=10$.

3. Плохая обусловленность
При сильной корреляции признаков (большом $\kappa$) требуется **большее** $l$ (например, 20–50), чтобы аппроксимировать мелкие собственные значения. На квадратичной функции с $\kappa=1000$ из лаб.1: при $l=5$ сходимость медленная, при $l=20$ – значительно быстрее.

4. Ускорение даже при малой истории
Уже при $l=1$ ускорение составляет:
- `abalone`: $4.16x$
- `a9a`: $10.71x$ 
При $l=5$: $6.12x$ и $11.26x$ соответственно. Минимальная история даёт многократный выигрыш по сравнению с градиентным спуском $l=0$.
